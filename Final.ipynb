```python
##task 3 merge data
sales = pd.merge( orders, order_items,on='order_id',how='inner')
display(sales.head())
```

```python
sales['total_price'] = (sales['quantity'] * sales['list_price']) - sales['discount']
sales
```

```python
##total price calculation
display(sales[['order_id', 'quantity','list_price', 'discount','total_price']].head())
```

```python
sales['order_date'] = pd.to_datetime( sales['order_date'],format='%d-%m-%Y')
print(sales['order_date'].dtype)
```

```python
##calculate recently
snapshot_date = (sales['order_date'].max() + pd.Timedelta(days=1))
print(snapshot_date)
recency = sales.groupby('customer_id').agg(last_order_date=('order_date', 'max'))
##number of days
recency['Recency'] =  (snapshot_date - recency['last_order_date']).dt.days
display(recency.head(10))
```

```python
frequency = sales.groupby('customer_id').agg(Frequency=('order_id', 'nunique'))
display(frequency.head(10))
```

```python
snapshot_date = (sales['order_date'].max() + pd.Timedelta(days=1))
print(snapshot_date)
```

```python
recency = sales.groupby('customer_id').agg(last_order_date=('order_date', 'max'))
recency
```

```python
recency['Recency'] = (snapshot_date - recency['last_order_date']).dt.days
display(recency.head(10))
```

```python
##calculate monetory
monetary = sales.groupby('customer_id').agg( Monetary=('total_price', 'sum'))
display(monetary.head(10))
```

```python
##combine recent,monetory,frequency
r = recency[['Recency']].join(frequency).join(monetary)
r = r.reset_index()
display(r.head(10))
```
